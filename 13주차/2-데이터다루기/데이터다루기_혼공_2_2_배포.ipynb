## 사이킷런으로 훈련 세트와 테스트 세트 나누기

```python
from sklearn.model_selection import train_test_split
```

```python
# 비율 생략시 기본인 약 7:3 으로 데이터 나눔

train_input, test_input, train_target, test_target = train_test_split(
    fish_data, fish_target, random_state=42)
```

```python
# 훈련 세트

print(train_input.shape, test_input.shape)
```

```python
# 테스트 세트

print(train_target.shape, test_target.shape)
```

```python
# 테스트 셋트 확인
# 13중 10개가 도미(1)이고 3개가 방어(0) 임
# 잘 섞인것 같으나 빙어의 비율이 조금 모자람 :
# 원래 도미와 빙어 비율 : 도미와 빙어의 비율 35/14 -> 2.5:1
# 테스트 셋트 비율 : 도미와 빙어의 비율 10/3 -> 3.3:1   ==> 샘플링 편향

print(test_target)       # 13 중 1 : 10개, 0 : 3개
print(len(test_target))
```

```python
# 샘플링 편향을 수정
# 클래스 비율에 맞게 데이터를 나눈다
# 훈련데이터가 작거나, 특정 클래스의 샘플개수가 적을 때 유용함

train_input, test_input, train_target, test_target = train_test_split(
    fish_data, fish_target, stratify=fish_target, random_state=42)
```

```python
print(test_target)
print(len(test_target))

# 원래 비율 : 도미와 빙어의 비율 36/14 -> 2.5:1
# 현재 비율 : 도미와 빙어의 비율 9/4   -> 2.25:1
```

##  수상한 도미 한마리

```python
from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier()

# 훈련 데이터로 훈련(학습)
kn.fit(train_input, train_target)

# 테스트 데이터로 성능 평가
kn.score(test_input, test_target)
```

```python
# 예측
# [길이, 무게] -> 길이 cm / 무게 g

print(kn.predict([ [25, 150] ]))

# 결과 : ???
```

## 기준을 맞춰라
- 데이터를 표현하는 기준이 다르면 알고리즘이 올바르게 예측할 수 없다. 거리기반일 경우 특히!
- 특정값을 일정한 기준으로 맞춘다 ==> 데이터 전처리

- 가장 널리 사용되는 전처리 방법 => 표준점수

```python
# 그래프 x축 y축 눈금 맞추기
# x축의 범위를 y축과 동일하게 (0~1000)

plt.scatter(train_input[:,0], train_input[:,1])
plt.scatter(25, 150, marker='^')
plt.scatter(train_input[indexes,0], train_input[indexes,1], marker='D')
plt.xlim((0, 1000))
plt.xlabel('length')
plt.ylabel('weight')
plt.show()
```

```python
# 위 그래프이면 x축은 영향이 미비하고 y축만 고려 대상이 된다

# 두 특성(길이와 무게)의 갑이 놓인 범위가 매우 다르다
# -> "두 특성의 스케일이 다르다" 라고 함
```

```python
train_input
```

```python
#  axis=0 : 각 열의 통계값을 계산

mean = np.mean(train_input, axis=0)  #  훈련데이터의 길이, 무게 -> 평균
std = np.std(train_input, axis=0)    #  훈련데이터의 길이, 무게 -> 표준편차
```

```python
print(mean, std)
```

```python
# 표준점수

# 표준점수 : 원본데이터에서 평균을 빼고 표준편자로 나눔
# 넘파이 배열은 기본적으로 브로드캐스팅을 적용
# train_scaled = (train_input 배열 - mean ) / std 배열

train_scaled = (train_input - mean) / std
```

## 전처리 데이터로 모델 훈련하기
- 표준 점수

```python
# 훈련 데이터 -> 표준점수
# 샘플 데이터는 표준점수 전 일반값이므로 --> 이상함!!!

plt.scatter(train_scaled[:,0], train_scaled[:,1])
plt.scatter(25, 150, marker='^')   # 표준점수 전 일반값
plt.xlabel('length')
plt.ylabel('weight')
plt.show()

# train_scaled 표준점수 , [25, 150] => 일반값이므로 이를 표준점수로 만든다
# 중요함!!
```

```python
# train_scaled 표준점수 , [25, 150] => 일반값이므로 이를 표준점수로 만든다

new = ([25, 150] - mean) / std
new
```

```python
plt.scatter(train_scaled[:,0], train_scaled[:,1])
plt.scatter(new[0], new[1], marker='^')  # 표준점수
plt.xlabel('length')
plt.ylabel('weight')
plt.show()
```

```python
# 훈련
kn.fit(train_scaled, train_target)
```

```python
# 테스트세트도 표준점수로 변환

test_scaled = (test_input - mean) / std
```

```python
# 평가
kn.score(test_scaled, test_target)
```

```python
print(kn.predict([new]))  # 예측을 위한 데이터 new

# 도미로 예측됨!
```

```python
# k-최근접 이웃 알고리즘으로 거리측정

distances, indexes = kn.kneighbors([new])
```

```python
# 가장 가까운 거리 5개

distances
```

```python
# 훈련 데이터의 indexes
indexes
```

```python
# 해당 위치에 있는 값의 타겟 값
print(train_target[indexes])
```

```python
#
plt.scatter(train_scaled[:,0], train_scaled[:,1])   # train 데이터
plt.scatter(new[0], new[1], marker='^')             # 측정 데이터

# 인접 위치 샘플
plt.scatter(train_scaled[indexes,0], train_scaled[indexes,1], marker='D')

plt.xlabel('length')
plt.ylabel('weight')
plt.show()

# 최근접 5개 모두가 도미이므로 -> (수상한 샘플)측정 데이터 도미!!
# 특성값의 스케일에 민감하지 않고 안정적인 예측할 수 있는 모델 완성!!
```
