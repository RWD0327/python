### 결측치 처리 : 삭제
- 데이터가 없는 행이나 열을 삭제
- dropna()
- dropna(how=?)
- dropna(thresh=?)

```python
import pandas as pd
import numpy as np

raw_data = {'first_name': ['Jason', np.nan, 'Tina', 'Jake', 'Amy'],
            'last_name': ['Miller', np.nan, 'Ali', 'Milner', 'Cooze'],
            'age': [42, np.nan, 36, 24, 73],
            'sex': ['m', np.nan, 'f', 'm', 'f'],
            'preTestScore': [4, np.nan, np.nan, 2, 3],
            'postTestScore': [25, np.nan, np.nan, 62, 70] }

# df = pd.DataFrame(raw_data, columns = ['first_name', 'last_name', 'age', 'sex', 'preTestScore', 'postTestScore'])
df = pd.DataFrame(raw_data)
df
```

```python
# 하나씩 실행해 보자
# df.isnull()
# df.isnull().sum()
# df.isnull().sum() / len(df)

df.isnull().sum() / len(df)
```

```python
# 결측치가 있는 행을 삭제

df.dropna()

# 결과 : ??
```

```python
# 위에서 삭제된 내용이 저장되지 않았다!
df
```

```python
# 삭제된 결과를 다른 객체에 저장

df_no_missing = df.dropna()
df_no_missing
```

```python
df.dropna(how='all')

# 결과 : 행의 모든 컬럼이 널이면 삭제
```

```python
df.dropna(how='any')

# 결과 : 행의 하나라도 널이면 삭제
```

```python
# thresh=1 :  결측치가 아닌 값이 최소 1개 이상 있는 행만 유지

df.dropna(thresh=1)
```

```python
# 결측치가 아닌 값이 5개 이상 있는 행만 유지

df.dropna(thresh=5)
```

- 범주형데이터의 갯수만큼 가변수를 생성 존재유무를 1 또는 0 으로 표현

```python
df = pd.DataFrame({'source': [0, 1, 2],
                   'target': [2, 2, 3],
                   'weight': [3, 4, 5],
                   'color' : ['red', 'blue', 'blue']})
df
```

```python
# 판다스 : get_dummies 사용
# 사이킷런 : OneHotEncoder, LabelEncoder 사용

#pd.get_dummies(df)

pd.get_dummies(df).astype(int)
```

### 범주형 데이터 처리
- 기수형 데이터 -> 명목형으로 변경 -> 원핫인코딩 적용

```python
df1 = pd.DataFrame({'source': [0, 1, 2],
                    'target': [2, 2, 3],
                    'weight': [3, 4, 5],
                    'color' : ['red', 'blue', 'blue']})
df1
```

```python
# weight 데이터는 숫자로 되어 있지만 연속형 데이터가 아니고 기수형 데이터다
# weight 데이터 -> 명목형으로 변경 -> 원핫인코딩 적용

weight_dict = {3:"M", 4:"L", 5:"XL"}
df1["weight_sign"] = df1["weight"].map(weight_dict)
df1
```

```python
# 혹시 True false 나오면 뒤에 astype(int) 해준다

#weight_sign = pd.get_dummies(df1["weight_sign"])
weight_sign = pd.get_dummies(df1["weight_sign"]).astype(int)
weight_sign
```

```python
# 두 데이터프레임 결합
# axis=1 좌우로 결합

pd.concat([df1, weight_sign], axis=1)
```

### 범주형 데이터 처리 : 바인딩
- 연속형 데이터 -> 범주형 데이터로 처리

```python
# 3개 부대 소속 군인들의 시험 전후 점수

raw_data = {'regiment': ['Nighthawks', 'Nighthawks', 'Nighthawks', 'Nighthawks', 'Dragoons', 'Dragoons', 'Dragoons', 'Dragoons', 'Scouts', 'Scouts', 'Scouts', 'Scouts'],
            'company': ['1st', '1st', '2nd', '2nd', '1st', '1st', '2nd', '2nd','1st', '1st', '2nd', '2nd'],
            'name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze', 'Jacon', 'Ryaner', 'Sone', 'Sloan', 'Piger', 'Riani', 'Ali'],
            'preTestScore': [4, 24, 31, 2, 3, 4, 24, 31, 2, 3, 2, 3],
            'postTestScore': [25, 94, 57, 62, 70, 25, 94, 57, 62, 70, 62, 70]}

# df = pd.DataFrame(raw_data)
df = pd.DataFrame(raw_data, columns = ['regiment', 'company', 'name', 'preTestScore', 'postTestScore'])
df
```

```python
# 연속형 데이터 -> 범주형 데이터로 변환
# cut 함수 사용
# 교재 227  : 범위 잘못 됨!!

bins = [0, 25, 50, 75, 100]  # bins 정의(0-25, 26-50, 51-75, 75-100)
group_names = ['Low', 'Okay', 'Good', 'Great']

df['categories'] = pd.cut(df['postTestScore'], bins, labels=group_names)

df
```

### 데이터 크기 맞추기 : 피쳐 스케일링
- 최대값-최대값 정규화 : 0~1 사이의 값으로 매핑하는 기법
- z-스코어 정규형 : 기존값을 표준 정규분포 값으로 변환하는 기법

```python
df = pd.DataFrame(
    {'A':[14.00,90.20,90.95,96.27,91.21],
     'B':[103.02,107.26,110.35,114.23,114.68],
     'C':['big','small','big','small','small']})

df
```

```python
# A열에 전체적으로 최소값을 빼준다

df["A"] - df["A"].min()
```

```python
# 공식에 그대로 대입
# 최소값-최대값 정규화

( df["A"] - df["A"].min() ) / (df["A"].max() - df["A"].min())
```

```python
( df["B"] - df["B"].min() ) / (df["B"].max() - df["B"].min())
```

```python
# z-스코어 정규형 : 기존값을 표준 정규분포 값으로 변환하는 기법

( df["B"] - df["B"].mean() ) / (df["B"].std() )
```
